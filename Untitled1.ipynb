# 📊 Credit Card Fraud Detection — Exploratory Data Analysis & Baseline Model


### 🔹 Project Overview
Fraudulent transactions cause huge losses for banks and users. In this notebook, we:
- Explore the **Kaggle Credit Card Fraud dataset** (anonymized features).
- Perform **EDA** on transaction amounts, time, and fraud imbalance.
- Train baseline models (**Logistic Regression & Random Forest**).
- Evaluate using **ROC-AUC** & **PR-AUC**, with threshold tuning.
- Package a **lightweight inference pipeline**.


---


Import Libraries

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (roc_auc_score, average_precision_score,
precision_recall_curve)
from joblib import dump
```

Load Dataset

```python
from google.colab import files
uploaded = files.upload()   # select creditcard.csv from your computer

import pandas as pd
df = pd.read_csv("creditcard.csv")
df.head()
```

Dataset Snapshot

```python
print("Dataset shape:", df.shape)
print("Class distribution:\n", df['Class'].value_counts())
print("Fraud percentage:", round(df['Class'].mean()*100, 4), "%")
```

Transaction Amount Distribution

```python
plt.figure(figsize=(8,5))
df['Amount'].hist(bins=50, color='skyblue')
plt.title('Transaction Amount Distribution')
plt.xlabel('Amount')
plt.ylabel('Count')
plt.show()
```

Transaction Time Distribution

```python
plt.figure(figsize=(8,5))
df['Time'].plot(kind='hist', bins=50, color='orange')
plt.title('Transaction Time Distribution')
plt.xlabel('Time (seconds)')
plt.ylabel('Count')
plt.show()
```

Train-Test Split

```python
X = df.drop(columns=['Class'])
y = df['Class']


X_train, X_valid, y_train, y_valid = train_test_split(
X, y, test_size=0.2, stratify=y, random_state=42
)
```

Baseline Models

```python
# Logistic Regression
logit_pipe = Pipeline([
('scaler', StandardScaler(with_mean=False)),
('clf', LogisticRegression(max_iter=200, class_weight='balanced'))
])


# Random Forest
rf_clf = RandomForestClassifier(
n_estimators=300,
class_weight='balanced_subsample',
n_jobs=-1,
random_state=42
)
```

Train Models

```python
logit_pipe.fit(X_train, y_train)
rf_clf.fit(X_train, y_train)
```

Evaluate Models

```python
proba_logit = logit_pipe.predict_proba(X_valid)[:,1]
proba_rf = rf_clf.predict_proba(X_valid)[:,1]


print("Logistic Regression ROC-AUC:", roc_auc_score(y_valid, proba_logit))
print("Logistic Regression PR-AUC:", average_precision_score(y_valid, proba_logit))
print("Random Forest ROC-AUC:", roc_auc_score(y_valid, proba_rf))
print("Random Forest PR-AUC:", average_precision_score(y_valid, proba_rf))
```

Threshold Tuning (Precision-Recall)

```python
prec, rec, thr = precision_recall_curve(y_valid, proba_rf)


# Targeting ~85% recall
target_recall = 0.85
idx = (np.abs(rec - target_recall)).argmin()
chosen_thr = thr[max(idx-1, 0)]


print("Chosen threshold:", round(float(chosen_thr),4))
```

Save Model Pipeline

```python
bundle = {
'model': rf_clf,
'threshold': float(chosen_thr),
'features': list(X.columns)
}


os.makedirs('models', exist_ok=True)
dump(bundle, 'models/baseline_pipeline.joblib')
print("Model saved → models/baseline_pipeline.joblib")
```

Key Takeaways

Dataset is highly imbalanced (~0.17% fraud).

Random Forest generally yields better PR-AUC than Logistic Regression.

Choosing the right threshold balances fraud recall vs false positives.
